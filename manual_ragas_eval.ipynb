```python
import pandas as pd
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Load results
df = pd.read_csv("evaluation_results.csv")

model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

faithfulness_scores = []
relevancy_scores = []

for _, row in df.iterrows():

    question = str(row["question"])
    expected = str(row["expected_answer"])
    generated = str(row["generated_answer"])

    # Answer Relevancy
    q_emb = model.encode(question)
    g_emb = model.encode(generated)

    relevancy = cosine_similarity(
        [q_emb],
        [g_emb]
    )[0][0]

    # Faithfulness approximation
    e_emb = model.encode(expected)

    faithfulness = cosine_similarity(
        [e_emb],
        [g_emb]
    )[0][0]

    relevancy_scores.append(relevancy)
    faithfulness_scores.append(faithfulness)

df["faithfulness"] = faithfulness_scores
df["answer_relevancy"] = relevancy_scores

print("=" * 50)

print(
    "Faithfulness:",
    round(np.mean(faithfulness_scores), 4)
)

print(
    "Answer Relevancy:",
    round(np.mean(relevancy_scores), 4)
)

print("=" * 50)
```
